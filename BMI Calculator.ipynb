BMI Calculator

```python
name = input('Enter your name')

weight = int(input( ' Enter the weight in pounds '))
height = int(input( ' Enter the height in inches ' ))
BMI = (weight * 703 ) / (height * height)
print(BMI)
```

```python
if BMI>0:
    if(BMI<18):
            print( name+" , you are underweight")
    elif(BMI<=24.9):
            print( name+" , you are normal. Continue excercising daily and improve your lifestsyle for even greater results.")
    elif(BMI<=29.9):
            print( name+" , you are overweight. You can get better by excercising 4x times a week")
    elif(BMI<=34.9):
            print( name+" , you are obese. You can get better by excercising 4x times a week ")
    elif(BMI<=39.9):
            print( name+" , you are morbidly obese. You can get better by excercising 4x times a week and a better diet")
    else:
        print(" You are severly obese. You can get better by excercising 4x times a week and a better diet ")

else:
    print(" Enter valid inputs ")
    


    


```
